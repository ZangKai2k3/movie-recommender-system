2. Thống kê mô tả các biến số (Mean, Min, Max, Quantile của Year, Runtime, Rating, Votes).
3. Phân tích phân bố số lượng phim qua các năm từ **2000 đến 2026** và vẽ biểu đồ cột.

6. Tìm ra Top 10 phim thịnh hành nhất và phân tích tương quan thời lượng - điểm số.

### 1. Khai báo thư viện & Nạp dữ liệu sạch

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# Thiết lập kích thước đồ thị mặc định
plt.rcParams['figure.figsize'] = (10, 5)
plt.rcParams['font.size'] = 11

df = pd.read_csv('data/interim/cleaned_movies_2000_2026.csv')
print(f'✅ Đã nạp thành công {len(df):,} bộ phim!')
```

### 2. Bảng Thống kê mô tả (Descriptive Statistics)

```python
print('--- THỐNG KÊ MÔ TẢ CÁC BIẾN SỐ CHÍNH ---')
display(df[['year', 'runtime_minutes', 'rating', 'votes']].describe().round(2))

print(f"Thời lượng trung bình: {df['runtime_minutes'].mean():.1f} phút")
print(f"Điểm đánh giá trung bình: {df['rating'].mean():.2f} / 10")
```

### 3. Phân bố số lượng phim theo năm (2000 - 2026)

```python
year_counts = df['year'].value_counts().sort_index()

# Vẽ biểu đồ cột
plt.figure(figsize=(12, 5))
plt.bar(year_counts.index, year_counts.values, color='#3b82f6', edgecolor='#1d4ed8')
plt.title('Phân Bố Số Lượng Phim Theo Năm Phát Hành (2000 - 2026)', fontsize=14, fontweight='bold')
plt.xlabel('Năm phát hành')
plt.ylabel('Số lượng phim')
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.xticks(year_counts.index, rotation=45)
plt.tight_layout()
plt.show()
```

### 4. Phân bố điểm số IMDb Rating

```python
# Vẽ biểu đồ histogram điểm số
plt.figure(figsize=(10, 5))
plt.hist(df['rating'], bins=30, color='#10b981', edgecolor='#047857')
plt.axvline(df['rating'].mean(), color='red', linestyle='dashed', linewidth=2, label=f'Điểm TB: {df["rating"].mean():.2f}')
plt.title('Phân Bố Điểm Đánh Giá IMDb Rating', fontsize=14, fontweight='bold')
plt.xlabel('Điểm số (Thang 1.0 - 10.0)')
plt.ylabel('Tần suất (Số phim)')
plt.legend()
plt.grid(axis='y', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 5. Top 15 Thể loại phim phổ biến nhất

```python
# Tách toàn bộ thể loại
all_genres = []
for item in df['genres'].dropna():
    all_genres.extend([g.strip() for g in item.split(',') if g.strip() and g.strip() != 'Unknown'])

genre_series = pd.Series(all_genres).value_counts().head(15)

# Vẽ biểu đồ thanh ngang
plt.figure(figsize=(10, 6))
plt.barh(genre_series.index[::-1], genre_series.values[::-1], color='#8b5cf6', edgecolor='#6d28d9')
plt.title('Top 15 Thể Loại Phim Xuất Hiện Nhiều Nhất (2000 - 2026)', fontsize=14, fontweight='bold')
plt.xlabel('Số phim gắn tag thể loại')
plt.grid(axis='x', linestyle='--', alpha=0.5)
plt.tight_layout()
plt.show()
```

### 6. Top 10 Bộ phim thịnh hành nhất (Nhiều lượt đánh giá nhất)

```python
top10_popular = df.sort_values(by='votes', ascending=False)[['title', 'year', 'genres', 'rating', 'votes']].head(10)
display(top10_popular)
print('\n👉 BƯỚC TIẾP THEO: Hãy mở file `04_Feature_Engineering_And_Recommendation.ipynb` để xây dựng mô hình gợi ý!')
```
